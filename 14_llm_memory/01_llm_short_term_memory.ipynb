```python
from langchain_google_genai import ChatGoogleGenerativeAI
```

```python
llm = ChatGoogleGenerativeAI(model="gemini-3.5-flash-lite")
```

```python
response = llm.invoke("Hi, My name is Saksham")
```

```python
print(response.content[0]["text"])
```

```python
response = llm.invoke("What is my name?")

print(response.content[0]["text"])
```

### Now use messages list to store conversations

```python
# create list of messages
messages = ["My name is Saksham"]
```

```python
output = llm.invoke(messages).content

print(output[0]["text"])
```

```python
# Append llm response to messages list
messages.append(output[0]["text"])

messages
```

```python
# Append user query in messages
messages.append("What is my name?")

messages
```

```python
output = llm.invoke(messages).content

print(output[0]["text"])
```
