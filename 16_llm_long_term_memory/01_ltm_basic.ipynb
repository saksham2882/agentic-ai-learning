```python
from langgraph.store.memory import InMemoryStore
```

```python
# creating a store
store = InMemoryStore()
```

```python
# creating a namespace
namespace = ("user", "u1")
```

### Creating Memories

```python
# Adding memories
store.put(namespace, "1", {"data": "User likes momos"})
store.put(namespace, "2", {"data": "User prefers dark mode"})
```

```python
# another namespace
namespace2 = ("user", "u2")

# Add memories
store.put(namespace2, "1", {"data": "User likes pizza"})
store.put(namespace2, "2", {"data": "User prefers grid style navigation"})
```

### Retrieving Memories

```python
store.get(namespace, "1")
```

### Retrieving all Memories

```python
items = store.search(namespace)

for item in items:
    print(item.value)
```

## Semantic Search

```python
from langchain_huggingface import HuggingFaceEmbeddings
from dotenv import load_dotenv
```

```python
load_dotenv()

embedding_model = HuggingFaceEmbeddings(model_name="sentence-transformers/all-MiniLM-L6-v2")
```

```python
store = InMemoryStore(index={'embed': embedding_model, 'dims':1536})
```

```python
namespace = ("users", "u1")
```

```python
# Adding memories
store.put(namespace, "1", {"data": "User lives in Delhi"})
store.put(namespace, "2", {"data": "User owns a pet dog"})
store.put(namespace, "3", {"data": "User is learning Agentic AI"})
store.put(namespace, "4", {"data": "User enjoys playing cricket on weekends"})
store.put(namespace, "5", {"data": "User prefers vegetarian food"})
store.put(namespace, "6", {"data": "User works as a software developer"})
store.put(namespace, "7", {"data": "User likes reading science fiction books"})
store.put(namespace, "8", {"data": "User wants to visit Japan"})
store.put(namespace, "9", {"data": "User uses Windows as their operating system"})
store.put(namespace, "10", {"data": "User enjoys listening to classical music"})
```

```python
items = store.search(namespace, query="what is user currently learning", limit=1)

for item in items:
    print(item.value)
```

```python
items = store.search(namespace, query="what are user's preferences", limit=3)

for item in items:
    print(item.value)
```
